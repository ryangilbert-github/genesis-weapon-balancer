# Genesis Module 1: The Physics Engine
## Part 2: Object-Oriented Simulation

**Objective:**
Now that we know *why* we need AI, we need to build the "World" it lives in.
We will create modular Python Classes to represent:
1. **The Weapon:** (Damage, Fire Rate, Mag Size)
2. **The Target:** (Health, Name)
3. **The Simulation:** A discrete-event loop that calculates "Time-To-Kill" (TTK).

**Key Skill:** Object-Oriented Programming (OOP) & Unit Testing.

```
import numpy as np
import pandas as pd

# 1. The Weapon Blueprint
class Weapon:
    def __init__(self, name, damage, fire_rate, mag_size):
        self.name = name
        self.damage = damage
        self.fire_rate = fire_rate
        self.mag_size = mag_size

    def get_dps(self):
        """Calculate raw Damage Per Second (ignoring reload)"""
        return self.damage * self.fire_rate

    def __repr__(self):
        return f"🔫 {self.name} | Dmg: {self.damage} | Rate: {self.fire_rate}/s | Mag: {self.mag_size}"

# 2. The Target Blueprint
class Monster:
    def __init__(self, name, hp):
        self.name = name
        self.hp = hp

    def __repr__(self):
        return f"👹 {self.name} (HP: {self.hp})"

# Unit Test: Create an AK-47 and a Goblin
ak47 = Weapon("AK-47", damage=30, fire_rate=10.0, mag_size=30)
goblin = Monster("Goblin", hp=100)

print(ak47)
print(goblin)
print(f"Raw DPS: {ak47.get_dps()}")
```

```
# 3. The Combat Logic (The Discriminator)
def simulate_ttk(weapon, monster):
    """
    Simulates a battle until the monster dies.
    Returns: Time in Seconds.
    """
    shots_needed = np.ceil(monster.hp / weapon.damage)

    # Reload Logic
    reloads = 0
    if shots_needed > weapon.mag_size:
        reloads = np.floor((shots_needed - 1) / weapon.mag_size)

    # Time Calculation
    # (Shots - 1) because the first bullet is fired at T=0.0s
    shooting_time = (shots_needed - 1) / weapon.fire_rate
    reload_time = reloads * 2.5 # Standard 2.5s reload animation

    total_time = shooting_time + reload_time
    return round(total_time, 4)

# TEST BENCH
# Let's test if the logic holds up against edge cases.

# Case A: One-Shot Kill (Should be 0.0s)
sniper = Weapon("AWP", damage=100, fire_rate=1.0, mag_size=5)
target = Monster("Weakling", hp=100)
print(f"Case A (One-Shot): {simulate_ttk(sniper, target)}s (Expected: 0.0)")

# Case B: Mag Dump (Exact mag size)
smg = Weapon("Vector", damage=10, fire_rate=10.0, mag_size=10)
target = Monster("Thug", hp=100) # Needs exactly 10 shots
print(f"Case B (Mag Dump): {simulate_ttk(smg, target)}s (Expected: 0.9)")

# Case C: Reload Required
pistol = Weapon("Glock", damage=10, fire_rate=5.0, mag_size=10)
target = Monster("Tank", hp=110) # Needs 11 shots (1 Reload)
print(f"Case C (Reload):   {simulate_ttk(pistol, target)}s (Expected > 2.5)")
```
